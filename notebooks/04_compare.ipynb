# 04 - compare encoders

**Any runtime (CPU is fine).** Reads every `probe_zones__*.csv` on Drive and
prints mIoU of the **class-balanced** linear probe on the validation glacier SI.
The `unet-scratch` rows are the U-Net from notebook 05, scored the same way
(`-48k` = 48,000 iterations, the reported one; the other = 12,000 iterations,
under-trained). They appear only after 05 has run.

Patch-level IoU (16 x 16 px majority labels, NA and padding excluded): good for
comparing encoders with each other, **not** comparable with the CaFFe paper's
pixel-level numbers (those come later, notebooks 05-07).

Always read band next to resolution: in SI, X-band = TSX at 7 m, L-band =
PALSAR at 17 m, so band and resolution are confounded. `RSAT@12m` vs
`RSAT@20m` is the one same-sensor, same-band resolution contrast (20 m = only 6
images).
